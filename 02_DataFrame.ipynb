## DataFrame 만들기

```python
import pandas as pd
```

```python
ls = ["a", "b", "c", "d", "e"]
ls_df = pd.DataFrame(ls, index=["ㄱ", "ㄴ", "ㄷ", "ㄹ", "ㅁ"], columns=["알파벳"])
ls_df
```

```python
dic = {"영수": 101, "영철": 102, "영삼": 103, "영자": 104, "영호": 105}
dic_df = pd.DataFrame(dic, index=["학번"])
dic_df
```

```python
dic_df = pd.DataFrame(dic.values())
dic_df.index = dic.keys()
dic_df.columns = ["성적"]
dic_df
```
