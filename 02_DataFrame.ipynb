## DataFrame 합치기
- `pd.concat([dataframe1, dataframe2], keys=[], names=[], axix=0|1, join="outer"|"inner")`
- index의 타입이 같아야 함

```python
df_samsung = pd.read_excel("/content/drive/MyDrive/financial pandas/samsung_naver_excel.xlsx", sheet_name="samsung", index_col=0)
df_samsung.head()
```

```python
df_naver = pd.read_excel("/content/drive/MyDrive/financial pandas/samsung_naver_excel.xlsx", sheet_name="naver", index_col=0)
df_naver.head()
```

```python
# axis=0
df = pd.concat([df_samsung, df_naver])
print(df.shape)
print(df_samsung.shape[0] + df_naver.shape[0] == df.shape[0])
```

```python
df = pd.concat([df_samsung, df_naver], keys=["naver", "samsung"], names=["stock", "date"])
df.head()
```

```python
print(df.index)
print(df.loc[[("naver", "2018-01-02")]])
```

```python
# axis=1, join="inner"

df = pd.concat([df_samsung, df_naver], keys=["naver", "samsung"], names=["stock", "date"], axis=1, join="inner")
df.head()
```

```python
# DataFrame 3개 합치기
# index 타입이 달라 inner join 해도 shape 0
df = pd.concat([df_samsung, df_naver, lotte_che_df], keys=["samsung", "naver", "lotte"], names=["stock", "date"], axis=1, join="inner")
print(f"df.shape: {df.shape}")
print(f"df_samsung.index: {type(df_samsung.index)}")
print(f"df_naver.index: {type(df_naver.index)}")
print(f"lotte_che_df.index: {type(lotte_che_df.index)}")
```

## NaN
- `df.isna()`
- `df.dropna()`
- `df.fillna(value)`

```python
df = pd.concat([df_samsung, df_naver, lotte_che_df], keys=["samsung", "naver", "lotte"], names=["stock", "date"], axis=1, join="outer")

df_samsung_close = df[[('samsung',  'Close')]]
df_samsung_close
```

```python
df_samsung_close.isna().sum()
```

```python
df_samsung_close.dropna().isna().sum()
```

```python
df_samsung_close.fillna(0).isna().sum()
```

```python
# NaN 값에 해당되는 날짜만 찾기
df_samsung_close[df_samsung_close.isna().values].index
```
