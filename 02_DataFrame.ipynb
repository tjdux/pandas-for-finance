## DataFrame 행과 열 추가하기

```python
#lotte_che = lotte_che_df[["Open", "High", "Low", "Close", "Volume"]]
lotte_che = lotte_che_df[lotte_che_df.columns[:-1]]
lotte_che.head()
```

```python
# 행 추가
lotte_che["Return"] = lotte_che["Close"].pct_change().fillna(0)
lotte_che.head()
```

```python
# 열 추가
lotte_che.loc["2021-12-23"] = [100, 200, 300, 400, 500, 600]
lotte_che.tail(1)
```

## DataFrame 행과 열 예제

```python
def return_fn(df):
  df_close = df[["Close"]]
  df_close["Return"] = df_close["Close"].pct_change().fillna(0)
  return df_close
```

```python
res = return_fn(lotte_che_df)
res.head()
```
